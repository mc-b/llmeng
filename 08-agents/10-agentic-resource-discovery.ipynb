## Referenzimplementierungen: Agentic Resource Discovery (ARD)

### Hugging Face Discover Tool

Das Hugging Face Discover Tool ermöglicht den Zugriff auf Tausende von Skills, ML-Anwendungen und MCP-Servern – auf Hugging Face – oder jedem anderen ARD-kompatiblen Dienst.

---

Einfach Suche um alle Agenten zu finden welche LLM Feintunen können

```python
%%bash
source ~/.hf/bin/activate
hf discover search "Fine tune a language model"
```

Suche MCP-Server welche Images erstellen können

```python
%%bash
source ~/.hf/bin/activate
hf discover search "Generate an image" --json --kind mcp
```

Gezielt eine Registry durchsuchen, hier die von Github

```python
%%bash
source ~/.hf/bin/activate
hf discover search "RAG" --registry-url https://agentfinder.github.com/api/v1
```
