KServe ist eine Kubernetes-native Plattform, mit der Machine-Learning- und LLM-Modelle standardisiert, skalierbar und API-basiert als Inference Services bereitgestellt werden.

---

### Installation

```python
%%bash
helm install kserve-crd oci://ghcr.io/kserve/charts/kserve-crd \
  --version v0.20.0 \
  --namespace kserve \
  --create-namespace

helm install kserve-resources oci://ghcr.io/kserve/charts/kserve-resources \
  --version v0.20.0 \
  --namespace kserve \
  --set kserve.controller.deploymentMode=Standard \
  --set kserve.controller.gateway.ingressGateway.className=traefik \
  --wait
```

Nach der Installation muss eine neue Custom Resource `InferenceService` vorhanden sein

```python
%%bash
kubectl explain InferenceService
```

Für die Verwendung von Hugging-Face-Modellen muss in KServe zusätzlich die passende `ClusterServingRuntime` installiert sein, da der `InferenceService` sonst keine Runtime für den Modelltyp `huggingface` findet.

```python
%%bash
helm install kserve-runtime-configs oci://ghcr.io/kserve/charts/kserve-runtime-configs \
  --version v0.20.0 \
  --namespace kserve \
  --set kserve.servingruntime.enabled=true
```

Ausgabe der Runtimes

```python
%%bash
kubectl get clusterservingruntimes
```

### LLM Model Deployen



```python
%%bash
kubectl create namespace kserve-test --dry-run=client -o yaml | kubectl apply -f -
kubectl apply -n kserve-test -f - <<EOF
apiVersion: serving.kserve.io/v1beta1
kind: InferenceService
metadata:
  name: qwen-llm
  annotations:
    serving.kserve.io/deploymentMode: RawDeployment
spec:
  predictor:
    minReplicas: 1
    model:
      modelFormat:
        name: huggingface
      args:
        - --model_name=qwen
        - --backend=huggingface
        - --dtype=float32
        - --max_model_len=1024
      storageUri: hf://Qwen/Qwen2.5-0.5B-Instruct
      resources:
        requests:
          cpu: "1"
          memory: 4Gi
        limits:
          cpu: "2"
          memory: 6Gi
EOF
```

Wir checken den Status

```python
%%bash
kubectl get inferenceservices qwen-llm -n kserve-test
kubectl get all -n kserve-test
```

Da wir den erstellen Service nicht ändern können, erstellen wir einen neuen

```python
%%bash
kubectl apply -n kserve-test -f - <<EOF
apiVersion: v1
kind: Service
metadata:
  name: qwen-llm-lb
spec:
  type: LoadBalancer
  selector:
    serving.kserve.io/inferenceservice: qwen-llm
  ports:
    - name: http
      port: 80
      targetPort: 8080
      protocol: TCP
EOF
```

### Testen

Über dessen NodePort können wir nun auf das LLM zugreifen

```python
%%bash
#source ~/data/env.py
cat <<EOF | tee ~/data/env-kserve-cpu.py
OPENAI_API_KEY="kserve-cpu"
HF_TOKEN=""
AI_KUBECONFIG="$AI_KUBECONFIG"
AI_MODEL="qwen"
AI_NAME=""
AI_IP="${AI_IP}"
AI_BASE_URL='http://localhost:$(kubectl get svc qwen-llm-lb -n kserve-test -o jsonpath='{.spec.ports[?(@.name=="http")].nodePort}')/openai/v1'
EOF
```

```python
%run ~/data/env-kserve-cpu.py
from openai import OpenAI

client = OpenAI(
    base_url=AI_BASE_URL,
    api_key="dummy",
)

response = client.chat.completions.create(
    model=AI_MODEL,
    messages=[
        {"role": "user", "content": "Antworte in einem Satz: Wer war John F. Kennedy?"}
    ],
    max_tokens=80,
    temperature=0.2,
)

print(response.choices[0].message.content)
```
