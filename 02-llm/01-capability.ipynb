## Gemeinsame Vorbereitung

Die nächste Zelle definiert eine Funktion zum Stellen von Fragen. Wir verwenden für beide Modelle **dieselbe Frage** und vergleichbare Einstellungen.


```python
from openai import OpenAI

client = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama", timeout=180.0)
MODELLE = {
    "Llama 3.1 (8B)": "llama3.1:8b-instruct-q4_K_M",
    "Gemma 3 (4B)": "gemma3:4b",
}

def fragen(modell, frage):
    antwort = client.chat.completions.create(
        model=modell,
        messages=[{"role": "user", "content": frage}],
        temperature=0.2,
        max_tokens=350,
    )
    return antwort.choices[0].message.content or "(Keine Textantwort)"
```

## 1. Fachwissen erklären

**Frage:** Können beide LLMs jemandem ohne Informatikvorkenntnisse verständlich erklären, was HTTP-Statuscodes bedeuten?

Achtet beim Vergleich auf **fachliche Korrektheit**, **Verständlichkeit** und **hilfreiche Beispiele**.


```python
frage = "Erkläre die HTTP-Statuscodes 200, 404 und 500 kurz für eine Person ohne Vorkenntnisse."

for name, modell in MODELLE.items():
    print(f"\n===== {name} =====")
    print(fragen(modell, frage))
```

## 2. Codevorschläge beurteilen

**Frage:** Kann ein LLM einen ersten Entwurf für einen Python-REST-Client liefern?

Vergleicht die Vorschläge: Ist die Anfrage erkennbar? Fehlen Fehlerbehandlung oder benötigte Pakete? **Würdet ihr den Code ohne Prüfung verwenden?**


```python
frage = "Schreibe ein kurzes Python-Beispiel für einen REST-Client."

for name, modell in MODELLE.items():
    print(f"\n===== {name} =====")
    print(fragen(modell, frage))
```
