## Funktionaler Überblick

Um uns einen funktionalen Überblick zu verschaffen, laden wir einige weitere LLMs aus der Registry und testen diese anschliessend mit verschiedenen Prompts.

Zuerst schauen wir uns die bereits geladenen LLMs an:


```python
%%bash
ollama list
```

Als weiteres Modell verwenden wir `smollm2:135m`, ein kompaktes Sprachmodell aus der SmolLM2-Familie mit 135 Millionen Parametern.

* Die **SmolLM-Modelle** stammen von **Hugging Face**, genauer vom Team **HuggingFaceTB** beziehungsweise Hugging Face Science / Training. Die SmolLM2-Arbeit wurde von Loubna Ben Allal und weiteren Autorinnen und Autoren bei Hugging Face veröffentlicht. ([Hugging Face][1])

Die **SmolLM-Familie** wird von **Hugging Face** entwickelt und veröffentlicht; sie gehört zu einer Reihe kompakter Sprachmodelle, die speziell für lokale und ressourcenschonende Ausführung konzipiert sind.

[1]: https://huggingface.co/collections/HuggingFaceTB/smollm2?utm_source=chatgpt.com "SmolLM2 - a HuggingFaceTB Collection"


```python
%%bash
ollama pull smollm2:135m
```

Setzen die Umgebung

```python
%%bash
source ~/data/env.py
cat <<EOF | tee ~/data/env.py
OPENAI_API_KEY="$OPENAI_API_KEY"
HF_TOKEN=""
AI_KUBECONFIG="$AI_KUBECONFIG"
AI_MODEL="smollm2:135m"
AI_NAME=""
AI_IP="${AI_IP}"
AI_BASE_URL='http://localhost:11434/v1'
EOF
```

Und stellen eine Anfrage, welches jedes Textmodell beantworten kann:

```python
%%ai
Wer war John F. Kennedy 
```
