## LMStudio

Der LM Studio Server ermöglicht es, lokale Sprachmodelle über eine OpenAI-kompatible API bereitzustellen und in eigene Anwendungen oder Workflows zu integrieren.

Gegenüber vielen anderen lokalen LLM-Tools ist LM Studio vor allem dann sinnvoll, wenn man schnell und unkompliziert ein Modell lokal testen und als API bereitstellen will. Die wichtigsten Vorteile sind:

* **Schneller Einstieg:** Kein eigener Inference-Server, keine komplexe Konfiguration, wenig Setup-Aufwand.
* **OpenAI-kompatible API:** Bestehende Tools oder Anwendungen, die OpenAI-APIs unterstützen, können oft mit kleinen Anpassungen angebunden werden.
* **Modellwechsel ohne grossen Aufwand:** Verschiedene lokale Modelle lassen sich relativ einfach vergleichen.
* **Gut für Prototyping:** Für Tests, Demos und erste Integrationen ist es schneller eingerichtet als viele produktionsnahe Alternativen.

### Hugging Face CLI

Falls der Modell-Download über LM Studio fehlschlägt, kann das Modell alternativ direkt über die Hugging Face CLI heruntergeladen und im LM-Studio-Modellverzeichnis abgelegt werden. Danach kann LM Studio das lokal vorhandene GGUF-Modell verwenden.

Die Schritte sind wie folgt:
* Installation Hugging Face CLI
* Download Models in LMStudio Model Verzeichnis


```python
%%bash
python3 -m venv ~/.hf-lmstudio
source ~/.hf-lmstudio/bin/activate
pip install -U huggingface_hub
```

```python
%%bash
source ~/.hf-lmstudio/bin/activate
hf download Qwen/Qwen3-Embedding-4B-GGUF --include "Qwen3-Embedding-4B-Q4_K_M.gguf" \
            --local-dir ~/.lmstudio/models/Qwen/Qwen3-Embedding-4B-GGUF/qwen3-embedding-4b

hf download Qwen/Qwen3-Embedding-8B-GGUF --include "Qwen3-Embedding-8B-Q4_K_M.gguf" \
            --local-dir /.lmstudio/models/Qwen/Qwen3-Embedding-8B-GGUF/qwen3-embedding-8b
```
