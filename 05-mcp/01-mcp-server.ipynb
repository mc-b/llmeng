# Model Context Protocol 

ist ein offenes Protokoll, mit dem Tools, Dateien, Repositories, Datenbanken oder APIs strukturiert an ein LLM angebunden werden.

Wir verwenden ein lokales LLM für die Textgenerierung; Kontext, Tools und Ressourcen kommen über MCP von aussen.

**Architektur**
* Das LLM ist mittels $AI_BASE_URL erreichbar
* Ein MCP-Client spricht MCP nach vorne
* Derselbe Client spricht nach hinten mit dem LLM über ein OpenAI-kompatibles API
* MCP-Server stellen Ressourcen bereit (Files, Git, SQL, HTTP, …)

```python
from mcp.server import MCPServer

mcp = MCPServer("webshop")

PRODUCTS = {
    "1": {"name": "Tastatur", "price": 79.0, "stock": 50},
    "2": {"name": "Maus", "price": 39.0, "stock": 50},
}

@mcp.resource("products://")
def list_products():
    return [{"id": k, **v} for k, v in PRODUCTS.items()]

@mcp.tool()
def buy_product(id: str, quantity: int):
    PRODUCTS[id]["stock"] -= quantity
    return "ok"

# main
await mcp.run_streamable_http_async(json_response=True, stateless_http=True)
```
