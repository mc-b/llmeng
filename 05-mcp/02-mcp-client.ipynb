# Model Context Protocol (Client)

Ein minimaler, aber vollständiger Beispiel-Client, der mit dem MCP-Server spricht (Resource + Tool),

**Wichtig:**
* Als erstes muss `mcp-server` gestartet werden. 


```python
import json
import pandas as pd
from IPython.display import display
from mcp.client.streamable_http import streamable_http_client
from mcp import ClientSession

async def main():
    async with streamable_http_client("http://localhost:8000/mcp") as (read, write):
        async with ClientSession(read, write) as session:
            await session.initialize()

            # MCP Resource lesen
            products = await session.read_resource("products://")

            # Rohform
            print("=== MCP Resource (Rohform) ===")
            print(products.model_dump_json(indent=2))

            # Tabellenform
            print("\n=== Produkte (Tabelle) ===")
            display(pd.DataFrame(json.loads(products.contents[0].text)))

            # MCP Tool aufrufen
            result = await session.call_tool(
                "buy_product",
                {"id": "1", "quantity": 1}
            )

            # Rohform
            print("\n=== MCP Tool Result (Rohform) ===")
            print(result.model_dump_json(indent=2))

            # Lesbare Ausgabe
            print("\n=== Ergebnis ===")
            print(result.content[0].text)

await main()
```
